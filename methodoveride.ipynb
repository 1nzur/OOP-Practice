```python
#Create a Shape class with an info() method that returns "I am a shape". Create Circle and Square children that override info() with their own messages.
class Shape:
    def info():
        print("I am a shape")
class Circle(Shape):
    def info():
        print("I am a circle")
class Square(Shape):
    def info():
        print("I am a square")
Shape.info()
Circle.info()
Square.info()
```

```python
#Create a Vehicle class with a fuel_type() method returning "Unknown". Override it in ElectricCar to return "Electric" and in GasCar to return "Gasoline".
class Vehicle:
    def fuel_type():
        return "Unknow"
class ElectricCar(Vehicle):
    def fuel_type():
        return "Electric"
class GasCar(Vehicle):
    def fuel_type():
        return "Gasoline"

print(Vehicle.fuel_type())
print(ElectricCar.fuel_type())
print(GasCar.fuel_type())
```

```python
#Create a Person class with a greet() method. Create a FormalPerson that does a partial override, calling super().greet() and adding "Pleased to meet you."
class Person:
    def __init__(self,name):
        self.name = name
    def greet(self):
        return("Welcome Samul")
class FormalPerson(Person):
    def __init__(self,name):
        super().__init__(name)

    def greet(self):
        base = super().greet()
        print(base,"Pleased to meet you")

p1 = FormalPerson("Samul")
p1.greet()
```
