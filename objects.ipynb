```python
#Create a Fruit class with a name attribute. Create three fruit objects and give each a different name.
class Fruit:
    name = ""
fruit1 = Fruit()
fruit2 = Fruit()
fruit3 = Fruit()
fruit1.name = "Banana"
fruit2.name = "Apple"
fruit3.name = "Mango"
print(fruit1.name , fruit2.name , fruit3.name)
```

```python
#Create a Phone class, create two objects, and use type() and isinstance() to verify their type.
class Phone:
    number = "98676767"

phone1 = Phone()
phone2 = Phone()
print(type(phone1))
print(isinstance(phone1,Phone))
```

```python
#Create a Color class with a name attribute. Create two objects, change one's name, and print both to confirm they are independent.
class Color:
    defaultcolor = "Red"

color1 = Color()
color2 = Color()
print(color1.defaultcolor , color2.defaultcolor)
color2.defaultcolor = "Blue"
print(color1.defaultcolor , color2.defaultcolor)
```
