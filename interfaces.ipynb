```python
#Create a Drawable interface with a draw() method. Implement it in Circle, Square, and Triangle classes
from abc import ABC,abstractmethod
class Drawable(ABC):
    @abstractmethod
    def draw(self):
        pass

class Circle(Drawable):
    def draw(self):
        print("Drawing Circle")
class Square(Drawable):
    def draw(self):
        print("Drawing Square")
class Triangle(Drawable):
    def draw(self):
        print("Drawing Triangle")

for i in [Circle(),Square(),Triangle()]:
    i.draw()
```

```python
#Create Readable and Writable interfaces. Create a TextFile class that implements both.
from abc import ABC,abstractmethod
class Readable(ABC):
    def read(self):
        pass

class Writable(ABC):
    def write(self):
        pass

class Textfile(Readable,Writable):
    def read(self):
        print("Reading text file")

    def write(self):
        print("Writting text file")

Textfile().read()
Textfile().write()
```

```python
#Design a Plugin interface with activate() and deactivate() methods. Create two different plugin classes that implement it.
from abc import ABC,abstractmethod
class Plugin(ABC):
    @abstractmethod
    def activate(self):
        pass

    @abstractmethod
    def deactivate(self):
        pass

class GamePlugin(Plugin):
    def activate(self):
        print("Activating game plugin")
    def deactivate(self):
        print("Deactivating game plugin")

class MusicPlugin(Plugin):
    def activate(self):
        print("Activating music plugin")
    def deactivate(self):
        print("Deactivating music plugin")

for i in [GamePlugin(),MusicPlugin()]:
    i.activate()
    i.deactivate()
```
