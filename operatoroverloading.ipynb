```python
#Create a Money class with __add__, __str__, and __eq__. Adding two Money objects should return a new one with the combined amounts.

class Money:
    def __init__(self,amount):
        self.amount = amount
    def __str__(self):
        return f"rs{self.amount}"
    def __add__(self,other):
        return (self.amount + other.amount)

    def __eq__(self,other):
        return(self.amount == other.amount)

m1 = Money(100)
m2 = Money(100)
print(m1+m2)
print(m1==m2)
print(m1)
```

```python
#Create a Cart class where len(cart) returns the number of items and print(cart) shows all items.

class Cart:
    def __init__(self,*items):
        self.items = list(items)

    def __str__(self):
        return str(self.items)

    def __len__(self):
        return len(self.items)

cart1 = Cart("Laptop","Mobile")
print(cart1)
print(len(cart1))
```

```python
#Create a Temperature class with __lt__, __eq__, and __str__. Create a list of temperatures and sort them.
class Temperature:
    def __init__(self, temp):
        self.temp = temp

    def __eq__(self, other):
        return self.temp == other.temp

    def __lt__(self, other):
        return self.temp < other.temp

    def __str__(self):
        return f"{self.temp}°C"


temps = [
    Temperature(30),
    Temperature(10),
    Temperature(50),
    Temperature(20),
    Temperature(40)
]

temps.sort()

print("Sorted temperatures:")
for t in temps:
    print(t)
```
