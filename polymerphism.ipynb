```python
#Create classes Car, Motorcycle, and Truck, each with a honk() method. Loop through a list of vehicles and call honk() on each.
class Car:
    def honk(self):
        return "car honks"
class Motorcycle:
    def honk(self):
        return "motorcycle honks"
class Truck:
    def honk(self):
        return "truck honks"

list1 = [Car(),Motorcycle(),Truck()]
for x in list1:
    print(x.honk())
```

```python
#Create classes Circle, Square, and Triangle, each with a describe() method that returns a string like "I am a circle." Write a function that accepts any shape and prints its description.

class Circle:
    def describe(self):
        print("I am a circle")
class Square:
    def describe(self):
        print("I am a square")
class Triangle:
    def describe(self):
        print("I am a triangle")
list1 = [Circle(),Square(),Triangle()]
for x in list1:
    x.describe()
```

```python
#Create classes CSVExporter, JSONExporter, and PDFExporter, each with an export(data) method that prints which format the data is being exported as.
class CSVExporter():
    def export(self,data):
        return (f"exporting {data} in CSV")

class JSONExporter():
    def export(self,data):
        return (f"exporting {data} in JSON")

class PDFExporter():
    def export(self,data):
        return (f"exporting {data} in PDF")

list1 = [CSVExporter(),JSONExporter(),PDFExporter()]
for x in list1:
    print(x.export("Samul"))
```
