```python
#Create an Employee class with a class variable company = "TechCorp" and instance variables name and role. Create three employees and print each one's company and name.
class Employee:
    company = "TechCorp"
    def __init__(self,name,role):
        self.name = name
        self.role = role
    def display(self):
        print(f"Company name = {self.company}, Name = {self.name}, Role = {self.role}")

e1 = Employee("Samul","HOR")
e2 = Employee("Kritesh","Janitor")
e3 = Employee("Kritagya","Guard Dai")
e1.display()
e2.display()
e3.display()
```

```python
#Add an employee_count class variable that increments each time an employee is created. Print the total after creating three employees.
class Employee:
    count = 0
    company = "TechCorp"
    def __init__(self,name,role):
        self.name = name
        self.role = role
        Employee.count += 1
    def display(self):
        print(f"Company name = {self.company}, Name = {self.name}, Role = {self.role}")

e1 = Employee("Samul","HOR")
e2 = Employee("Kritesh","Janitor")
e3 = Employee("Kritagya","Guard Dai")
e1.display()
e2.display()
e3.display()
print(Employee.count)
```
