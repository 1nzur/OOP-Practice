```python
#Create a Book class. Create two book objects and give each a title, author, and pages attribute. Print them all.
class Book:
    title = ""
    author = ""
    pages = ""

book1 = Book()
book1.title = "book1"
book1.author = "book1author"
book1.pages = 99
book2 = Book()
book2.title = "book2"
book2.author = "book2author"
book2.pages = 999
print(book1.title, book1.author, book1.pages)
print(book2.title, book2.author, book2.pages)


```

```python
#Create a Pet class, create an object, set its name and age, then modify the age and print before and after.
class Pet:
    name = ""
    age = 0
pet1 = Pet()
pet1.name = "Max"
pet1.age = 10

print(pet1.name , pet1.age)
pet1.age = 12
print(pet1.name , pet1.age)
```

```python
#Create a Product class with two objects. Give each a name, price, and quantity. Print the total value (price times quantity) for each.
class Product:
    name = ""
    price = 0
    quantity = 0

product1 = Product()
product1.name = "Laptop"
product1.price = 180000
product1.quantity = 1

product2 = Product()
product2.name = "Mobile"
product2.price = 50000
product2.quantity = 2

print(f"Product name : {product1.name} Quantity = {product1.quantity} Price = {product1.price} Total = {product1.quantity * product1.price}")
print(f"Product name : {product2.name} Quantity = {product2.quantity} Price = {product2.price} Total = {product2.quantity * product2.price}")
```
