```python
#Extend the BankAccount class with a transfer(other_account, amount) method that withdraws from one account and deposits into another.
class BankAccount:
    __total_accounts = 0  
    def __init__(self, owner, balance=0):
        self.__owner = owner
        self.__balance = balance
        self.__transactions = []
        BankAccount.__total_accounts += 1

    @property
    def balance(self):
        return self.__balance

    @property
    def owner(self):
        return self.__owner

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            self.__transactions.append(f"+${amount:.2f}")
            return True
        return False

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            self.__transactions.append(f"-${amount:.2f}")
            return True
        return False

    def get_statement(self):
        return self.__transactions.copy()

    @classmethod
    def get_total_accounts(cls):
        return cls.__total_accounts

    def transfer(self,otheracc,amount):
        if amount<self.__balance:
            self.__balance -= amount
            otheracc.__balance += amount
            print(f"sucessfully debited rs {amount} to another acc balance for acc1 = {self.__balance} balance for acc2 = {otheracc.__balance}")

acc1 = BankAccount("Alice", 1000)
acc2 = BankAccount("Bob", 300)
acc1.transfer(acc2,500)
```

```python
#Create a Library class that has a private list of books. Add methods to add a book, remove a book, and search by title. The book list should never be directly accessible.
class Library:
    def __init__(self,booklist):
        self.__booklist = booklist

    def add(self,book):
        if book in self.__booklist:
            print("Book already in library")
        else:
            self.__booklist.append(book)
            print("Book added sucessfully")
            print(self.__booklist)

    def remove(self,book):
        if book in self.__booklist:
            self.__booklist.remove(book)
            print("Book removed sucessfully")
            print(self.__booklist)
        else:
            print("Book not in booklist")

    def search(self,book):
        if book in self.__booklist:
            print("Book found in booklist")
        else:
            print("Book is not in booklist")
lib1 = Library(['Book1','Book2','Book3','Book4','Book5'])
lib1.add("Book6")
lib1.search("Book6")
lib1.remove("Book6")
lib1.add("Book1")
lib1.search("Book100")
lib1.remove("Book100")
```

```python
#Create an Inventory class with private stock counts per item. Add methods to add stock, sell items (with validation that stock does not go below zero), and a read-only property for total items in stock.
class Inventory:
    __total = 0
    def __init__(self,item,count):
        self.__item = item
        self.__count = count
        Inventory.__total += 1

    def addstock(self,amount):
        if amount>0:
            self.__count += amount
            print(f"Restocked {self.__item} now total count for that item = {self.__count}")
        else:
            print("Invalid input")
    def sellitem(self,amount):
        if amount > self.__count:
            print("The stock cannout go below 0")
        else:
            self.__count -= amount
            print(f"Sucessfully sold {amount} of {self.__item} new stock = {self.__count}")
    @classmethod
    def total(cls):
        return cls.__total
item1 = Inventory("Laptop",10)
item2 = Inventory("Mobile",3)
item1.addstock(10)
item2.sellitem(3)
item2.sellitem(100)
print(Inventory.total())
```
