```python
#Create Walkable and Talkable classes. Then create a Robot class that inherits from both. Test all methods.
class Walkable:
    def walk(self):
        print("Can walk")

class Talkable:
    def talk(self):
        print("Can talk")

class Robot(Walkable,Talkable):
    pass

Robot().walk()
Robot().talk()
```

```python
#Create a TimestampMixin (adds created_at) and ValidateMixin (adds validate()). Combine them in a Post class.

class TimestampMixin:
    def created_at(self):
        print("Created today")

class ValidateMixin:
    def validate(self):
        print("Validated today")

class Post(TimestampMixin,ValidateMixin):
    pass

Post().created_at()
Post().validate()
```

```python
#Create a class that inherits from 3 parents. Print its __mro__ and predict which method gets called when parents share a method name.

class A:
    def speak(self):
        print("A speaks")

class B:
    def speak(self):
        print("B speaks")

class C:
    def speak(self):
        print("C speaks")

class D(A,B,C):
    pass

d1 = D()
d1.speak()
print(D.__mro__)
```
