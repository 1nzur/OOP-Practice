```python
#Create a Password class that stores a private password. Add a check_password(guess) method that returns True or False, but never reveals the actual password.
class Password:
    def __init__(self,password):
        self.__password = password

    def checkpassword(self,guess):
        return(self.__password == guess)

pw1 = Password("Samul@123")
print(pw1.checkpassword("xd"))
print(pw1.checkpassword("Samul@123"))

```

```python
#Create a Temperature class with a private __celsius attribute. Add a set_temperature() method that only accepts values between -273.15 and 1000.
class Temperature:
    def __init__(self,celsius=0):
        self.__celsius = celsius

    def set_temperature(self,newvalue):
        if -273.15<=newvalue<=1000:
            self.__celsius = newvalue
            print(f"New temperature = {self.__celsius}")
        else:
            print("Invalid temperature")
t1 = Temperature()
t1.set_temperature(100)
t1.set_temperature(-300)

    
```

```python
#Take the unprotected BankAccount example and add a withdraw() method that prevents the balance from going negative.
class BankAccount:
    def __init__(self,balance):
        self.__balance = balance

    def withdraw(self,amount):
        if amount <= self.__balance:
            self.__balance -= amount
            print(f"Sucessfully withdrawed rs{amount} your new balance is {self.__balance}")
        else:
            print("Amount should not be greater than balance")
acc1 = BankAccount(10950)
acc1.withdraw(10950)
acc1.withdraw(900)
```
