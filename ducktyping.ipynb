```python
#Create three unrelated classes: ConsoleLogger, FileLogger, and CloudLogger. Each should have a log(message) method. Write a function that accepts any logger and uses it.
class ConsoleLogger:
    def log(self,message):
        return f"Console logger: {message}"
class FileLogger:
    def log(self,message):
        return f"File logger: {message}"
class CloudLogger:
    def log(self,message):
        return f"Cloud Logger {message}"

def log(thing,message):
    print(thing.log(message))

for i in [ConsoleLogger(),FileLogger(),CloudLogger()]:
    print(i.log("Samul"))
```

```python
#Create classes Circle, Button, and TextBox -- each with a draw() method. Write a render_all(items) function that loops through and draws each one.

class Circle:
    def draw(self):
        return "drawing circle"
class Button:
    def draw(self):
        return "drawing button"
class TextBox:
    def draw(self):
        return "drawing textbox"

def render_all(items):
    for item in items:
        print(item.draw())

render_all([Circle(), Button(), TextBox()])
```

```python
#Write a function safe_speak(thing) that uses try/except to call thing.speak() and prints a friendly error if the method does not exist.

class Robot:
    def speak(self):
        return "talks"
class Human:
    def speak(self):
        return "talks"
class Dog:
    pass


def safe_speak(*things):
    for thing in things:
        try:
            print(thing.speak())
        except:
            print("No Speak function")
safe_speak(Robot(),Dog(),Human())
```
