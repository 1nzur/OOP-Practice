```python
#Create a Restaurant class with attributes (name, cuisine, rating) and methods (serve_food, get_review). Create two restaurant objects.

class Restaurant:
    def __init__(self,name,cuisine,rating):
        self.name = name
        self.cuisine = cuisine
        self.rating = rating
    def serve_food(self):
        print(f"Now serving {self.cuisine}")
    def get_review(self):
        print(f"Rating: {self.rating}")

r1 = Restaurant("Ozone","MOMO",0)
r2 = Restaurant("Oxygen","Pizza",5)
r1.serve_food()
r1.get_review()
r2.serve_food()
r2.get_review()
```

```python
#Create a Pet class with attributes (name, species, energy) and methods (play, sleep, eat). Make the energy go down when playing and up when eating or sleeping.
class Pet:
    def __init__(self,name,species,energy=50):
        self.name = name
        self.species = species
        self.energy = energy

    def play(self):
        self.energy = self.energy - 10
        print(f"Pet is now playing new energy = {self.energy}")

    def eat(self):
        self.energy = self.energy + 10
        print(f"Pet is now eating new energy = {self.energy}")

    def sleep(self):
        self.energy = self.energy + 20
        print(f"Pet is now sleeping new energy = {self.energy}")

dog1 = Pet("Leo","husky",40)
dog1.play()
dog1.play()
dog1.sleep()
dog1.play()
dog1.eat()
```
