```python
#Rewrite the Shape example from Lesson 6.1 using ABC and @abstractmethod. Add Circle, Rectangle, and Triangle subclasses.

from abc import ABC,abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius

    def area(self):
        return 3.14*self.radius**2

class Rectangle(Shape):
    def __init__(self,width,height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

class Triangle(Shape):
    def __init__(self,base,height):
        self.base = base
        self.height = height

    def area(self):
        return 0.5*self.base*self.height

for i in [Circle(7),Rectangle(10,10),Triangle(10,10)]:
    print(i.area())
```

```python
#Create an abstract Employee class with abstract method calculate_salary() and concrete method display_info(). Create FullTime and PartTime subclasses.
from abc import ABC,abstractmethod
class Employee(ABC):
    def __init__(self,name):
        self.name = name
    @abstractmethod
    def calculate_salary(self,hours):
        pass

    def display_info(self):
        return f"Your name is {self.name} and you are {self.__class__.__name__} "
class FullTime(Employee):
    def __init__(self,name,salary):
        super().__init__(name)
        self.salary = salary

    def calculate_salary(self,hours):
        print("Your total salary is",self.salary * hours)
class PartTime(Employee):
    def __init__(self,name,salary):
        super().__init__(name)
        self.salary = salary

    def calculate_salary(self,hours):
        print("Your total salary is",self.salary * hours)

emp1 = FullTime("Samul",10000)
print(emp1.display_info())
emp2 = PartTime("Inzur",5000)
print(emp2.display_info())
emp1.calculate_salary(10)
emp2.calculate_salary(5)
```

```python
#Build an abstract Notifier with abstract method send(message). Create EmailNotifier, SMSNotifier, and PushNotifier subclasses.
from abc import ABC,abstractmethod
class Notifier(ABC):
    @abstractmethod
    def send(self,message):
        pass

class EmailNotifier(Notifier):
    def send(self,message):
        return f"Message: {message} recieved via email"

class SMSNotifier(Notifier):
    def send(self,message):
        return f"Message: {message} recieved via SMS"

class PushNotifier(Notifier):
    def send(self,message):
        return f"Message: {message} recieved via Push Notification"

for i in [EmailNotifier(),SMSNotifier(),PushNotifier()]:
    print(i.send("Hello"))
```
