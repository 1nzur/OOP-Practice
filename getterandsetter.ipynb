```python
#Create a Temperature class with a @property for celsius that rejects values below -273.15 (absolute zero).
class Temperature:
    def __init__(self,celsius=500):
        self.__celsius = celsius

    @property
    def celcius(self):
        return self.__celsius
    @celcius.setter
    def celcius(self,newvalue):
        if -273.15<=newvalue<=1000:
            self.__celsius = newvalue
            print(f"New temperature = {self.__celsius}")
        else:
            print("Invalid temperature")
t1 = Temperature()
print(t1.celcius)
t1.celcius = 100
t1.celcius = -300
```

```python
#Create an Employee class with a read-only employee_id property and a settable salary property that only accepts positive values.
class Employee:
    def __init__(self,employee_id,salary):
        self.__employee_id = employee_id
        self.__salary = salary
    @property
    def employee_id(self):
        return self.__employee_id
    @property
    def salary(self):
        return self.__salary

    @salary.setter
    def salary(self,value):
        if value>0:
            self.__salary = value
            print(f"New salary is {self.__salary}")
        else:
            print("Salary cannot be less than 0")
e1 = Employee(1,500)
print(e1.employee_id)
e1.salary = 100
e1.salary = -300
```

```python
#Create a Rectangle class where width and height have setters (positive values only), and area is a read-only computed property.
class Rectangle:
    def __init__(self,width,height):
        self.__width = width
        self.__height = height
    @property
    def width(self):
        return(self.__width)

    @width.setter
    def width(self,newvalue):
        if newvalue>0:
            self.__width = newvalue
        else:
            print("Enter positive value")
    @property
    def height(self):
        return(self.__height)

    @height.setter
    def height(self,newvalue):
        if newvalue>0:
            self.__height = newvalue
        else:
            print("Enter positive value")

    @property
    def area(self):
        return self.__height * self.__width
rec1 = Rectangle(10,10)
print("Area of rectangle =" , rec1.area)
rec1.height = 20
rec1.height = 20
print("Area of rectangle =" , rec1.area)
rec1.height = -30
```
