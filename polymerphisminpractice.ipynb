```python
#Extend the Notification System with a SlackNotification class. Override send() to print "Sending SLACK to #channel: message". Add it to the notifications list and verify it works without changing any existing code.
class Notification:
    def __init__(self, recipient, message):
        self.recipient = recipient
        self.message = message

    def send(self):
        raise NotImplementedError("Subclasses must implement send()")

    def __str__(self):
        return f"To: {self.recipient} | {self.message}"

class EmailNotification(Notification):
    def send(self):
        print(f"Sending EMAIL to {self.recipient}: {self.message}")

class SMSNotification(Notification):
    def send(self):
        print(f"Sending SMS to {self.recipient}: {self.message}")

class PushNotification(Notification):
    def send(self):
        print(f"Sending PUSH to {self.recipient}: {self.message}")

class SlackNotification(Notification):
    def send(self):
        print(f"Sending SLACK to #channel: {self.message}")


notifications = [
    EmailNotification("[email protected]", "Your order shipped!"),
    SMSNotification("+1234567890", "Your OTP is 4521"),
    PushNotification("bob_device", "New message received"),
    SlackNotification("SlackMessage","Slack Message is sent")
]


for n in notifications:
    n.send()  
    print(n)   
```

```python
class Vehicle:
    fuel_price = 197
    
    def __init__(self, mileage):
        self.mileage = mileage

    def fuel_cost(self, distance):
        raise NotImplementedError("Subclasses must implement fuel_cost()")
    
    def __str__(self):
        return f"Milage of {self.__class__.__name__} is {self.mileage}"
    
    def __lt__(self, other):
        return self.mileage < other.mileage


class Car(Vehicle):
    def fuel_cost(self, distance):
        return (distance / self.mileage) * self.fuel_price


class Truck(Vehicle):
    def fuel_cost(self, distance):
        return (distance / self.mileage) * self.fuel_price


class Motorcycle(Vehicle):
    def fuel_cost(self, distance):
        return (distance / self.mileage) * self.fuel_price


vehicles = [Car(30), Truck(20), Motorcycle(40)]

def calculate_fleet_cost(vehicles, distance):
    total = 0
    for vehicle in vehicles:
        cost = vehicle.fuel_cost(distance)
        total += cost
        print(f"{vehicle.__class__.__name__}: Rs.{cost:.2f}")
    print(f"Total fleet cost: Rs.{total:.2f}")
    return total


calculate_fleet_cost(vehicles, 100)
```

```python
#Build a complete Shape system with Circle, Rectangle, and Triangle. Include area(), perimeter(), __str__, __add__ (combine areas), and __lt__ (compare by area). Sort a list of mixed shapes by area.

class Shape:
    def area(self):
        raise NotImplementedError("Each subclass must have their own area")
    def perimeter(self):
        raise NotImplementedError("Each subclass must have their own perimeter")
    def __str__(self):
        return f"Area of {self.__class__.__name__} is {self.area()}"

    def __add__(self,other):
        return(self.area() +other.area())
        
    def __lt__(self,other):
        return self.area() < other.area()

class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius

    def area(self):
        return 3.14*self.radius*self.radius
    def perimeter(self):
        return 2*3.14*self.radius
class Rectangle(Shape):
    def __init__(self,width,height):
        self.width = width
        self.height = height

    def area(self):
        return self.width*self.height

    def perimeter(self):
        return 2*(self.width+self.height)
class Triangle(Shape):
    def __init__(self,base,height):
        self.base = base
        self.height = height

    def area(self):
        return 0.5*self.base*self.height

def area(thing):
    print(thing.area())

for i in [Circle(7),Rectangle(10,10),Triangle(10,10)]:
    area(i)

Rectangle(10,10)+Triangle(10,10)
print(Rectangle(10,10))
```
