```python
#Create a Book class with an __init__ that takes title, author, and pages. Create two books and print their details.
class Book:
    def __init__(self,title,author,page):
        self.title = title
        self.author = author
        self.page = page

    def details(self):
        print(f"Title: {self.title}, Author: {self.author}, Page: {self.page}")

book1 = Book("Book1","Author1",67)
book2 = Book("Book2", "Author2", 420)
book1.details()
book2.details()



```

```python
#Create a Car class where color defaults to "White". Create one car with a custom color and one without.
class Car:
    def __init__(self,company,color = "white"):
        self.company = company 
        self.color = color
car1 = Car("Hundai","Red")
car2 = Car("Toyota")
print(car1.color)
print(car2.color)
```

```python
#Create a Rectangle class with width and height in __init__, and an area() method that returns the area.
class Rectangle:
    def __init__(self,width,height):
        self.width = width
        self.height = height
    def area(self):
        print(f"Area = {self.width * self.height}")
rec1 = Rectangle(10,10)
rec1.area()
```
