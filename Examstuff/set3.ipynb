```python
#3. Vehicle Rental System
#A rental company rents:
#Cars 
#Bikes 
#Trucks 
#Each vehicle has different rental rates and insurance calculations.
#Task:
#Design a system where rental calculations are handled through OOP principles.

from abc import ABC,abstractmethod
class Vehicle(ABC):
    def __init__(self,rent_rate,insurance):
        self.rent_rate = rent_rate
        self.insurance = insurance
    
    @abstractmethod
    def rent(self):
        pass

class Car(Vehicle):
    def rent(self):
        print(f"To rent a car you need to pay {self.rent_rate} and your insurance is {self.insurance}")
class Bike(Vehicle):
    def rent(self):
        print(f"To rent a bike you need to pay {self.rent_rate} and your insurance is {self.insurance}")
class Truck(Vehicle):
    def rent(self):
        print(f"To rent a car you need to pay {self.rent_rate} and your insurance is {self.insurance}")

for i in [Car(2000,100000),Bike(1500,150000),Truck(5000,500000)]:
    i.rent()
```

```python
#10. Tax Calculation System
#The software calculates taxes for:
#Individuals 
#Small Businesses 
#Corporations 
#Each follows different tax rules.
#Task:
#Create an extensible tax calculation module.

from abc import ABC,abstractmethod
class Tax(ABC):
    def __init__(self,income):
        self.income = income

    @abstractmethod
    def taxcalc(self):
        pass
class Individuals(Tax):
    def taxcalc(self):
        print(f"Tax for Individuals is 10% so your tax is {self.income * 0.1}")
class SmallBusiness(Tax):
    def taxcalc(self):
        print(f"Tax for SmallBusiness is 15% so your tax is {self.income * 0.15}")
class Corporations(Tax):
    def taxcalc(self):
        print(f"Tax for Corporations is 20% so your tax is {self.income * 0.2}")

for i in [Individuals(10000),SmallBusiness(50000),Corporations(100000)]:
    i.taxcalc()
```

```python
#11. Ticket Booking System
#The platform sells:
#Movie Tickets 
#Bus Tickets 
#Flight Tickets 
#Each has different cancellation policies.
#Task:
#Design a system that handles booking and cancellation efficiently.
from abc import ABC,abstractmethod
class Ticket(ABC):
    def __init__(self,price):
        self.price = price 
    @abstractmethod
    def booking(self):
        pass
    @abstractmethod
    def cancel(self):
        pass

class MovieTicket(Ticket):
    def booking(self):
        print(f"Your movie ticket has been booked price = {self.price}")
    def cancel(self):
        print(f"You have cancelled your movie ticket and have been refunded {self.price}")

class BusTicket(Ticket):
    def booking(self):
        print(f"Your bus ticket has been booked price = {self.price}")
    def cancel(self):
        print(f"You have cancelled your bus ticket and have been refunded {self.price}")

class FlightTicket(Ticket):
    def booking(self):
        print(f"Your flight ticket has been booked price = {self.price}")
    def cancel(self):
        print(f"You have cancelled your flight ticket and have been refunded {self.price}")
for i in [MovieTicket(400),BusTicket(1100),FlightTicket(22000)]:
    i.booking()
    i.cancel()
```

```python
#12. Notification System
#Notifications can be sent through:
#Email 
#SMS 
#Push Notification 
#Future channels:
#WhatsApp 
#Slack 
#Task:
# Design the notification service.
from abc import ABC,abstractmethod
class Notification(ABC):
    def __init__(self,sender,message):
        self.sender = sender
        self.message = message
    @abstractmethod
    def send(self):
        pass
class Email(Notification):
    def send(self):
        return f"Message: \"{self.message}\" recieved via email from {self.sender}"
class SMS(Notification):
    def send(self):
        return f"Message: \"{self.message}\" recieved via SMS from {self.sender}"
class Whatsapp(Notification):
    def send(self):
        return f"Message: \"{self.message}\" recieved via Whatsapp from {self.sender}"

def send_msg(sent):
    print(sent.send())

for i in [Email("Samul","Hi"),SMS("Samul","Hi"),Whatsapp("Samul","Hi")]:
    send_msg(i)
```

```python
#14. Hospital Appointment System
#Appointments can be:
#General Consultation 
#Specialist Consultation 
#Emergency Appointment 
#Each has different booking fees and rules.
#Task:
#Design a scalable appointment system.

from abc import ABC,abstractmethod

class HospitalSys(ABC):
    def __init__(self,price):
        self.price = price

    @abstractmethod
    def book(self):
        pass
class General(HospitalSys):
    def book(self):
        print(f"Booking must be done before hand and the price is {self.price}")
class Specialist(HospitalSys):
    def book(self):
        print(f"Booking must be done through the specific specialist and price is {self.price}")
class Emergency(HospitalSys):
    def book(self):
        print(f"Booking can be done on the spot and price is {self.price}")

for i in [General(1000),Specialist(1500),Emergency(3000)]:
    i.book()
```

```python
#15. Warehouse Inventory Management
#Inventory contains:
#Electronics 
#Clothing 
#Furniture 
#Each category has different storage and shipping requirements.
#Task:
#Design an inventory management solution using OOP.

from abc import ABC,abstractmethod
class Inventory(ABC):
    @abstractmethod
    def storage(self):
        pass
    @abstractmethod
    def ship(self):
        pass
class Electronics(Inventory):
    def storage(self):
        print("Electronics are stored in the electronics section which is kept away from water so no water damages occour")
    def ship(self):
        print("Shipping electonics cost extra free of 100$")
class Clothing(Inventory):
    def storage(self):
        print("Clothing are stored in the clothing section which is kept away from water so no water damages occour")
    def ship(self):
        print("Shipping cloathing cost extra free of 20$")
class Furniture(Inventory):
    def storage(self):
        print("Furniture are stored in the electronics section which is kept away from water so no water damages occour")
    def ship(self):
        print("Shipping furniture cost extra free of 5$")

for i in [Electronics(),Clothing(),Furniture()]:
    i.storage()
    i.ship()
```

```python
#16. Multi-Country Commission Management System
#You are building an international student recruitment platform.
#Universities define commission rates. 
#Agents receive a percentage of that commission. 
#Different countries have different commission structures. 
#Future commission rules will vary by country and university. 
#Task:
#Design a commission calculation engine that supports future rule changes without modifying existing code.

from abc import ABC,abstractmethod
class Comission(ABC):
    def __init__(self,uni):
        self.uni = uni
    @abstractmethod
    def calc(self):
        pass
class USA(Comission):
    def calc(self):
        if self.uni == "coventry":
            amount = 1000
            rate = 0.1
            print(f"Agent will get commission of {amount * rate}")
        elif self.uni == "stanford":
            amount = 1500
            rate = 0.09
            print(f"Agent will get commission of {amount * rate}")
        else:
            print("Uni not in our list")
class Nepal(Comission):
    def calc(self):
        if self.uni == "coventry":
            amount = 198000
            rate = 0.1
            print(f"Agent will get commission of {amount * rate}")
        elif self.uni == "stanford":
            amount = 100000
            rate = 0.05
            print(f"Agent will get commission of {amount * rate}")
        else:
            print("Uni not in our list")

for i in  [USA("coventry"),Nepal("stanford")]:
    i.calc()
```

```python
#17. Ride Sharing Platform (Uber/Pathao)
#The platform supports:
#Bike Rides 
#Car Rides 
#Premium Rides 
#Pricing depends on:
#Distance 
#Surge Pricing 
#Time of Day 
#Future:
#Electric Vehicle Rides 
#Shared Rides 
#Task:
# Design the fare calculation system.

from abc import ABC,abstractmethod
class Ride(ABC):
    def __init__(self,distance,price,time):
        self.distance = distance
        self.price = price
        self.time = time
    @abstractmethod
    def calc(self):
        pass

class Bike(Ride):
    def calc(self):
        if self.time == "office time":
            self.price = self.price + 100
            print(f"Since your ride is a bike ride and it is in office time your ride will cost extra rs100 so your total cost will be {self.price * self.distance}")
        elif self.time == "other hours":
            print(f"Since your ride is a bike ride and it is not in office time your ride will have no extra cost and your total cost will be {self.price * self.distance}")
        else:
            print("Put either office time or other hours")

class Car(Ride):
    def calc(self):
        if self.time == "office time":
            self.price = self.price + 300
            print(f"Since your ride is a car ride and it is in office time your ride will cost extra rs300 so your total cost will be {self.price * self.distance}")
        elif self.time == "other hours":
            print(f"Since your ride is a car ride and it is not in office time your ride will have no extra cost and your total cost will be {self.price * self.distance}")
        else:
            print("Put either office time or other hours")
class Premium(Ride):
    def calc(self):
        if self.time == "office time":
            self.price = self.price + 500
            print(f"Since your ride is a premium ride and it is in office time your ride will cost extra rs500 so your total cost will be {self.price * self.distance}")
        elif self.time == "other hours":
            print(f"Since your ride is a car ride and it is not in office time your ride will have no extra cost and your total cost will be {self.price * self.distance}")
        else:
            print("Put either office time or other hours")

for i in [Bike(10,18,"office time"),Car(10,30,"other hours"),Premium(10,50,"9am")]:
    i.calc()    
```
