```python
#You are building a notification module for an application. The system needs to support multiple communication channels (Email,SMS) but the core service logic should not need to change everytime a new channel (like WhatsApp or Push) is added.
class Notification:
    def __init__(self,sender,message):
        self.sender = sender
        self.message = message
class Email(Notification):
    def send(self):
        return f"Message: \"{self.message}\" recieved via email from {self.sender}"
class SMS(Notification):
    def send(self):
        return f"Message: \"{self.message}\" recieved via SMS from {self.sender}"
class Whatsapp(Notification):
    def send(self):
        return f"Message: \"{self.message}\" recieved via Whatsapp from {self.sender}"

def send_msg(sent):
    print(sent.send())

for i in [Email("Samul","Hi"),SMS("Samul","Hi"),Whatsapp("Samul","Hi")]:
    send_msg(i)
```

```python
#2. Shipping Cost Calculator System
#You are designing a shipping service module for an online store.
#Currently, the system supports:
#Standard Shipping
#Express Shipping
#Later, it may include:
#Drone Delivery
#International Priority Shipping
#Same-Day Delivery
#Task:
#Design a system where shipping cost calculation logic can be extended without modifying existing code.

class StandardShipping:
    def __init__(self,price):
        self.price = price
    def ship(self):
        return f"For {self.__class__.__name__} total price = Rs{self.price}"
class ExpressShipping:
    def __init__(self,price):
        self.price = price
    def ship(self):
        return f"For {self.__class__.__name__} total price = Rs{self.price}"

def ship(method):
    print(method.ship())

for i in [StandardShipping(100),ExpressShipping(180)]:
    ship(i)
```

```python
#3. Logging System for an Application
#You are building a logging framework for a large application.
#Initially, logs are written to:
#Console
#File
#Later, logs may also go to:
#Database
#Cloud services (AWS CloudWatch, Azure Monitor)

#Task:
#Design a system where adding new log destinations does not require modifying existing logger code.

#3. Logging System for an Application
#You are building a logging framework for a large application.
#Initially, logs are written to:
#Console
#File
#Later, logs may also go to:
#Database
#Cloud services (AWS CloudWatch, Azure Monitor)

#Task:
#Design a system where adding new log destinations does not require modifying existing logger code.

class Console:
    def log(self):
        return f"Logs are written to console"
class File:
    def log(self):
        return f"Logs are written to File"

def log(file):
    print(file.log())

for i in [Console(),File()]:
    log(i)
```
