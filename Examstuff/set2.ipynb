```python
#An e-commerce company sells products with different discount policies.
#Design an abstract class Product with:
#private field productName
#constructor
# abstract method calculateDiscount()
# concrete method displayProduct()
# Create a subclass Electronics that gives 10% discount.

from abc import ABC,abstractmethod

class Product(ABC):
    def __init__(self,name,price):
        self.__name = name
        self.__price = price

    @abstractmethod
    def calculateDiscount(self):
        pass

    def displayProduct(self):
        print(f"Product Name = {self.__name} and Price = {self.__price}")

    @property
    def name(self):
        return self.__name

    @property
    def price(self):
        return self.__price
    

class Electronics(Product):
    def calculateDiscount(self):
        print(f"Your product is {self.name} which is an electronic so you get a 10% discount and your new price is {self.price - (self.price * 0.10)}")

p1 = Electronics("Laptop",100000)
p1.displayProduct()
p1.calculateDiscount()
```

```python
#A transport company offers multiple vehicle services.
# Design an abstract class Vehicle with:
#private field vehicleNumber
#constructor
#abstract method calculateFare(double distance)
#concrete method displayVehicleInfo()
#Create a subclass Taxi that charges $2 per kilometer.

from abc import ABC,abstractmethod
class Vehicle(ABC):
    def __init__(self,vehicleNumber):
        self.__vehicleNumber = vehicleNumber

    @abstractmethod
    def calculateFare(self,distance):
        pass

    def displayVehicleInfo(self):
        print(f"Your vehicle number is {self.__vehicleNumber}")

    @property
    def vehicleNumber(self):
        return self.__vehicleNumber

class Taxi(Vehicle):
    def calculateFare(self, distance):
        print(f"Your vehicle number is {self.vehicleNumber} and your fare for {distance} is {2*distance}$")

t1 = Taxi(6767)
t1.calculateFare(100)
```

```python
#A university evaluates different student categories.
#Design an abstract class Student with:
#private field studentName
#constructor
#abstract method calculateScholarship()
#concrete method displayStudentInfo()
#Create a subclass UndergraduateStudent that receives a scholarship based on GPA.
from abc import ABC,abstractmethod
class Student(ABC):
    def __init__(self,name):
        self.__name = name

    @abstractmethod
    def calculateScholorship(self,gpa):
        pass

    def displayStudentInfo(self):
        print(f"Student Name: {self.__name}")

    @property
    def name(self):
        return self.__name

class UndergraduateStudent(Student):
    def calculateScholorship(self, gpa):
        if gpa == 4:
            print(f"{self.name} has recieved 90% Scholorship")
        elif 4 > gpa >= 3.6:
            print(f"{self.name} has recieved 80% Scholorship")
        else:
            print(f"{self.name} has recieved No scholorship")

s1 = UndergraduateStudent("Samul")
s1.calculateScholorship(4)
s1.displayStudentInfo()

s2 = UndergraduateStudent("Inzur")
s2.calculateScholorship(3.5)
```

```python
#A company wants a basic payroll system. Design an abstract class
#Employee with a private field name, a constructor to initialize it, an
#abstract method calculateSalary(), and a concrete method
#displayPaySlip(). Create a subclass FullTimeEmployee that calculates
#salary using a fixed monthly salary. Write a Main class to test this scenario.

from abc import ABC,abstractmethod
class Employee(ABC):
    def __init__(self,name):
        self.__name = name

    @abstractmethod
    def calculateSalary(self):
        pass
    def displayPaySlip(self):
        print(f"Name = {self.__name} and you are a full timer")

    @property
    def name(self):
        return self.__name
    
class FullTimeEmployee(Employee):
    salary = 10000
    def calculateSalary(self):
        print(f"Name = {self.name} and monthly salary is {self.salary}")

class Main:
    def run(self):
        FullTimeEmployee("Samul").displayPaySlip()
        FullTimeEmployee("Samul").calculateSalary()

Main().run()
```

```python
#A Student should not allow direct modification of marks.
#Task:
#Design a class Student with:
#private attribute __marks
#method add_marks(int value)
#method get_marks()
#Requirement:
#Marks should only increase through add_marks()
#Direct access to __marks must not be allowed

class Student:
    def __init__(self,name,marks):
        self.name = name
        self.__marks = marks

    def get_marks(self):
        print(self.__marks)

    def add_marks(self,value):
        if self.__marks + int(value) > 100 or int(value)<0:
            print("Invalid input")
        else:
            self.__marks += value
            print("Marks added")
s1 = Student("Samul",70)
s1.get_marks()
s1.add_marks(20)
s1.get_marks()
s1.add_marks(100)

```

```python
#An ATM should protect user balance from direct access.
#Task:
#Design a class ATMAccount with:
#private attribute __balance
#method deposit(amount)
#method withdraw(amount)
#method get_balance()

class ATMAccount:
    def __init__(self,balance):
        self.__balance = balance

    def deposite(self,amount):
        if amount<=0:
            print("Invalid")
        else:
            self.__balance += amount
            print(f"Sucessfully deposited {amount}, your new balance is {self.__balance}")
    def withdraw(self,amount):
        if amount > self.__balance or amount<0:
            print("Insufficient funds")
        else:
            self.__balance -= amount
            print(f"Sucesffuly withdrawed {amount}, your new balance is {self.__balance}")

    def get_balance(self):
        print(f"Your balance is {self.__balance}")

acc1 = ATMAccount(500)
acc1.get_balance()
acc1.deposite(500)
acc1.get_balance()
acc1.withdraw(300)
acc1.get_balance()
```
