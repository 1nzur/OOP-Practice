```python
#A base class Device has a method operate(). You have a Light and a Thermostat.
#Task: Override operate() so that Light prints "Glowing" and Thermostat prints "Adjusting Temperature". 

class Device():
    def operate(self):
        print("Operating")
class Lights(Device):
    def operate(self):
        print("Glowing")
class Thermostat(Device):
    def operate(self):
        print("Adjusting Temperature")

d = Device()
l = Lights()
t = Thermostat()
d.operate()
l.operate()
t.operate()
```

```python
#Create a class hierarchy Person → Student → GraduateStudent.
#Task: Use super() to initialize attributes across all levels. 

class Person():
    def __init__(self,name):
        self.name = name

class Student(Person):
    def __init__(self,name,age):
        super().__init__(name)
        self.age = age

class GraduateStudent(Student):
    def __init__(self,name,age,year):
        super().__init__(name,age)
        self.year = year

a = GraduateStudent("Samul",18,2022)
print(a.name)
```

```python
#Create an abstract class Shape with an abstract method calculateArea(). Implement subclasses Circle and Rectangle. 
from abc import ABC,abstractmethod
class Shape(ABC):
    @abstractmethod
    def calculateArea(self):
        pass
class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
    def calculateArea(self):
        return 3.14*self.radius**2
class Rectangle(Shape):
    def __init__(self,length,breadth):
        self.length = length
        self.breadth = breadth

    def calculateArea(self):
        return self.length * self.breadth

for i in [Circle(7),Rectangle(10,10)]:
    print(i.calculateArea())
```

```python
#A company wants a basic payroll system. Design an abstract class Employee with a private field name, a constructor to initialize it, an abstract method calculateSalary(), and a concrete method displayPaySlip(). Create a subclass FullTimeEmployee that calculates salary using a fixed monthly salary. Write a Main class to test this scenario. 
from abc import ABC,abstractmethod
class Employee(ABC):
    def __init__(self,name):
        self.__name = name

    @abstractmethod
    def calculateSalary(self):
        pass
    def displayPaySlip(self):
        print(f"Name = {self.__name} and you are a full timer")

    @property
    def name(self):
        return self.__name
    
class FullTimeEmployee(Employee):
    salary = 10000
    def calculateSalary(self):
        print(f"Name = {self.name} and monthly salary is {self.salary}")

class Main:
    def run(self):
        FullTimeEmployee("Samul").displayPaySlip()
        FullTimeEmployee("Samul").calculateSalary()

Main().run()
```

```python
#Create a class BankAccount with private fields accountNumber and balance.
#Task: Demonstrate encapsulation by providing appropriate getter and setter methods. 

class BankAccount():
    def __init__(self,accountNumber,balance):
        self.__accountNumber = accountNumber
        self.__balance = balance

    @property
    def accountNumber(self):
        return self.__accountNumber

    @property
    def balance(self):
        return self.__balance
    @balance.setter
    def balance(self,amount):
        if amount<0:
            print("invalid")
        else:
            self.__balance = amount
            print(f"New balance = {self.__balance}")
acc1 = BankAccount(1,100)
print(acc1.accountNumber)
acc1.balance = 300
acc1.balance = -10
```

```python
#Create an abstract class Animal with an abstract method makeSound(). Implement subclasses Dog and Cat. 
from abc import ABC,abstractmethod
class Animal(ABC):
    abstractmethod
    def makeSound(self):
        pass
class Dog(Animal):
    def makeSound(self):
        print("Bark")
class Cat(Animal):
    def makeSound(self):
        print("Meow")

for i in [Dog(),Cat()]:
    i.makeSound()
```

```python
#Smartphone inherits from MobilePhone, which inherits from Electronics.
#Task: Demonstrate the use of super() to initialize attributes across all levels. 

class Electronics:
    def __init__(self,company):
        self.company = company
class MobilePhone(Electronics):
    def __init__(self,company,color):
        super().__init__(company)
        self.color = color
class SmartPhone(MobilePhone):
    def __init__(self, company, color, model):
        super().__init__(company, color)
        self.model = model

m1 = SmartPhone("Samsung","Red","E-1")
print(m1.company,m1.color,m1.model)
```

```python
#Create a base class Employee and derive classes Manager and Developer.
#Task: Demonstrate runtime polymorphism by overriding a method work(). 

class Employee:
    def work(self):
        print("Employee Must work")
class Manager(Employee):
    def work(self):
        print("Manager needs to manage the company")
class Developer(Employee):
    def work(self):
        print("Developer needs to develop the system")

for i in [Employee(),Manager(),Developer()]:
    i.work()
```

```python
#Design an abstract class Account with an abstract method calculateInterest(). Implement subclasses SavingsAccount and CurrentAccount. 

from abc import ABC,abstractmethod
class Account(ABC):
    def __init__(self,balance):
        self.balance = balance
    @abstractmethod
    def calculateIntrest(self):
        pass
class SavingsAccount(Account):
    def calculateIntrest(self):
        print(f"Saving accounts have an intrest of 4.5% and your intrest comes out to {self.balance * 0.045}")
class CurrentAccount(Account):
    def calculateIntrest(self):
        print(f"Current accounts have an intrest of 2% and your intrest comes out to {self.balance * 0.02}")

for i in [SavingsAccount(100000),CurrentAccount(100000)]:
    i.calculateIntrest()
```
