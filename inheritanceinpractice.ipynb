```python
#Extend the Employee system: add an Intern class with a university attribute and override get_info().
class Employee:
    def __init__(self, name, emp_id, salary):
        self.name = name
        self.emp_id = emp_id
        self.salary = salary

    def get_info(self):
        return f"{self.name} (ID: {self.emp_id}) - ${self.salary:,.2f}"

class Manager(Employee):
    def __init__(self, name, emp_id, salary, department):
        super().__init__(name, emp_id, salary)
        self.department = department
        self.team = []

    def add_to_team(self, employee):
        self.team.append(employee)

    def get_info(self):
        base = super().get_info()
        return f"{base}\n  Department: {self.department}, Team: {len(self.team)}"

class Developer(Employee):
    def __init__(self, name, emp_id, salary, language):
        super().__init__(name, emp_id, salary)
        self.language = language

    def get_info(self):
        base = super().get_info()
        return f"{base}\n  Language: {self.language}"
class Intern(Employee):
    def __init__(self,name, emp_id,salary,university):
        super().__init__(name,emp_id,salary)
        self.university = university
    def get_info(self):
        return(f"Name: {self.name}, University: {self.university}")

dev1 = Developer("Alice", "D001", 85000, "Python")
dev2 = Developer("Bob", "D002", 90000, "JavaScript")
mgr = Manager("Carol", "M001", 110000, "Engineering")
intern = Intern("Samul","I101",50000,"Coventry")


mgr.add_to_team(dev1)
mgr.add_to_team(dev2)


staff = [dev1, dev2, mgr, intern]
for person in staff:
    print(person.get_info())
    print()
```

```python
#Build a School system: create a Person base with Student and Teacher children. Each has a unique role() method.
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age

    def role(self):
        return ("Role of human being")

class Student(Person):
    def __init__(self,name,age,grade):
        super().__init__(name,age)
        self.grade = grade

    def role(self):
        return (f"Role of student is to study {self.grade}")
class Teacher(Person):
    def __init__(self,name,age,subject):
        super().__init__(name,age)
        self.subject = subject

    def role(self):
        return (f"Your role is to teach {self.subject}")

list1 = [Person("Samul",18), Student("Samul",18,"Bachlors"),Teacher("Samul",18,"Math")]
for x in list1:
    print(x.role())
```
