```python
#Single inheritance
class A:
    pass
class B(A):
    pass
```

```python
#Multilevel inheritance
class A:
    pass
class B(A):
    pass
class C(B):
    pass
```

```python
#Hierarchical inheritance
class A:
    pass
class B(A):
    pass
class C(A):
    pass
```

```python
#Multiple
class A:
    pass
class B:
    pass
class C(A,B):
    pass
```

```python
#Create a multilevel chain: Vehicle to Car to ElectricCar. Each level adds one method. Test that ElectricCar can call all three.

class Vehicle:
    def engine():
        print("It Has an engine")
class Car(Vehicle):
    def type():
        print("It is a car")
class ElectricCar(Car):
    def electric():
        print("It is electric")

ElectricCar.electric()
ElectricCar.type()
ElectricCar.engine()

```

```python
#Create hierarchical inheritance: Shape as parent, with Circle, Rectangle, and Triangle as children. Each child has its own area() method.
class Shape:
    def area(self):
        print("Area")
class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius

    def area(self):
        print(f"Area of circle is {self.radius * self.radius}")
class Rectangle(Shape):
    def __init__(self,length,width):
        self.length = length 
        self.width = width
    def area(self):
        print(f"Area of rectangle is {self.length * self.width}")
class Triangle(Shape):
    def __init__(self,base,height):
        self.base = base
        self.height = height

    def area(self):
        print(f"Area of triagnel is {0.5 * self.base * self.height}")
shapes = [Shape(),Rectangle(10,10),Circle(7),Triangle(10,10)]
for shape in shapes:
    shape.area()
```
